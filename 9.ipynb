```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import DBSCAN
from sklearn.preprocessing import StandardScaler
df = pd.read_csv(r"C:\Users\HP\Downloads\9.make_moons.csv")
```

```python
df.head()
```

```python
X = df[["Feature_1", "Feature_2"]]
X = StandardScaler().fit_transform(X)
```

```python
model = DBSCAN(eps=0.5, min_samples=10)
df["Cluster"] = model.fit_predict(X)
```

```python
df.head()

clusters = df["Cluster"].nunique()
if -1 in df["Cluster"].values:
    clusters -= 1

print("Number of Clusters:", clusters)
print("Noise Points:", list(df["Cluster"]).count(-1))

plt.scatter(X[:, 0], X[:, 1], c=df["Cluster"], cmap="summer")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("DBSCAN Clustering")
plt.show()
```

```python
print("Total Core Points:", len(model.core_sample_indices_))
```
